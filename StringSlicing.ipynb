### String Slicing 
### (Forward or Backward)

```python
''' ------------------------------------
Brute force method
For backward slicling, just reverse the string and follow the same steps
-------------------------------------'''
inputString = "ABC"
def forwardStringSlicing(inputString):   
    finalList = []      
    for IndexInputString in range(len(inputString)):
        tempList = []
        for subString in finalList:
            # Post appending inputString[IndexInputString]
            tempList.append(subString + inputString[IndexInputString])
        finalList = finalList + list(inputString[IndexInputString]) + tempList
    print(len(finalList))        
    print(finalList)
forwardStringSlicing(inputString)
```

```python
inputString = "ABC"
''' ------------------------------------
Recursive Solution with helper function. 
forwardStringSlicingHelper function does the recursive part
-------------------------------------'''
def forwardStringSlicing(inputString): 
    finalList = []
    forwardStringSlicingHelper(inputString, 0, finalList)
    print(len(finalList))        
    print(finalList)

def forwardStringSlicingHelper(inputString, index, finalList): 
    if not inputString:
        return
    tempList = []
    for subString in finalList:
        # Post appending inputString[IndexInputString]
        tempList.append(subString + inputString[index])
    finalList = finalList + list(inputString[index]) + tempList    
    inputString = inputString[1:]
    forwardStringSlicingHelper(inputString[], index+1, finalList)

forwardStringSlicing(inputString)
```
