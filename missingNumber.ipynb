```python
def missingNumber(nums):
    n = len(nums)
    # Create a dictionary with keys and values from 0 to n
    my_dict = {i: 0 for i in range(n + 1)}
    for num in nums:
        if num in my_dict.keys():
            my_dict[num] += 1
    print(my_dict)
    # Find and return the first missing number (with value 0)
    for key, value in my_dict.items():
        if value == 0:
            return key
nums = [9,6,4,2,3,5,7,0,1]        
print(missingNumber(nums))
```

```python
def missingNumber(nums):
    num_set = set(nums)
    n = len(nums) + 1
    for number in range(n):
        if number not in num_set:
            return number
nums = [9,6,4,2,3,5,7,0,1]        
print(missingNumber(nums))        
```
